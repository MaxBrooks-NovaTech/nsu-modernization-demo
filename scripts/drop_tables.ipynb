```python
# Drop the PascalCase fact tables from the attached NSU_DEMO Lakehouse.
# This does not access Files/ and does not call notebookutils.fs.exists().

fact_tables = [
    "FactEnrollmoent",
    "CertificationCatalog",
    "CensusEnrollment",
    "RecruitmentFunnel",
    "LineageSummary"
]

for table_name in fact_tables:
    print(f"Dropping table if it exists: {table_name}")
    spark.sql(f"DROP TABLE IF EXISTS `{table_name}`")

print("PascalCase fact-table cleanup complete.")

# Verify that the tables no longer exist in the Spark catalog.
remaining = [
    table_name
    for table_name in fact_tables
    if spark.catalog.tableExists(table_name)
]

if remaining:
    raise RuntimeError(f"These tables still exist: {remaining}")

print("Verified: no PascalCase fact tables remain.")
```
